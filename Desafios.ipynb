```python
senha_correta = "Admin@123"
tentativas = 0
limite_tentativas = 3

while tentativas < limite_tentativas:
    senha_digitada = input("Digite sua senha de acesso: ")

    if senha_digitada == senha_correta:
        print("Login efetuado com sucesso! Bem-vindo ao sistema.")
        # Força a saída do laço quando acerta
        break
    else:
        tentativas += 1
        tentativas_restantes = limite_tentativas - tentativas

        if tentativas_restantes > 0:
            print(f"Senha incorreta. Você ainda tem {tentativas_restantes} tentativa(s).\n")
        else:
            print("ALERTA DE SEGURANÇA: Conta bloqueada por excesso de tentativas falhas!")
```

```python
total_arrecadado = 0.0
quantidade_vendas = 0
valor_venda = -1.0  # Iniciamos com -1 apenas para entrar no loop

print("--- SISTEMA DE CAIXA INICIADO ---")
print("Digite 0 quando quiser encerrar o expediente.\n")

while valor_venda != 0:
    valor_venda = float(input("Digite o valor da venda: R$ "))

    if valor_venda != 0:
        total_arrecadado += valor_venda
        quantidade_vendas += 1

# Quando o usuário digita 0, o loop quebra e vem direto para cá
print("\n--- FECHAMENTO DO CAIXA ---")
print(f"Total de itens vendidos: {quantidade_vendas}")
print(f"Faturamento total: R$ {total_arrecadado:.2f}")
```

### Desafio 3

```python
logs_do_sistema = [
    "ERRO: Banco de dados desconectado as 03:00",
    {"produto": "Placa de Video RTX", "quantidade": 1, "preco": 3500.00, "status": "ok"},
    "   mouse gamer  ,  5 ,  45.90   ",
    "TESTE: admin, 1, 0.00",
    {"status": "erro", "mensagem": "Timeout no gateway de pagamento"},
    "Monitor Ultrawide; 1; 1500.00",
    {"produto": "Pendrive 32GB", "quantidade": 3, "preco": 25.00, "status": "ok"},
    "Cadeira de Escritorio, 2"
]
```

```python
for log in logs_do_sistema:
  produto = ""
  quantidade = 0
  preco = 0.0

  # EXTRAÇÃO E NORMALIZAÇÃO

  if type(log) == dict:
      # Tratando dados do Sistema Novo (API / Dicionário)
      if log["status"] == "erro":
          print(f"Alerta [Novo Sistema]: {log['mensagem']}")
          print("-" * 40)
          continue
      else:
          produto = log["produto"].title()
          quantidade = int(log["quantidade"])
          preco = float(log["preco"])

  elif type(log) == str:
      # Tratando dados do Sistema Antigo (Texto / String)
      if log.startswith("ERRO") or log.startswith("TESTE"):
          print(f"-> Ignorado [Sistema Antigo]: Log de erro ou teste.")
          print("-" * 40)
          continue
      else:
          log_padronizado = log.replace(";", ",")
          dados = log_padronizado.split(",")

          if len(dados) != 3:
              print(f"Venda Corrompida: Faltam dados. Log: '{log}'")
              print("-" * 40)
              continue

          produto = dados[0].strip().title()
          quantidade = int(dados[1].strip())
          preco = float(dados[2].strip())
  else:
      print("Formato de dado desconhecido.")
      continue

  # REGRA DE PRECIFICAÇÃO

  subtotal = quantidade * preco

  if subtotal > 1000:
      categoria = "Venda VIP"
      valor_final = subtotal * 0.90 # 10% de desconto
  elif subtotal >= 100:
      categoria = "Venda Padrão"
      valor_final = subtotal # Sem alteração
  else:
      categoria = "Venda Varejo"
      valor_final = subtotal + 15.00 # Taxa de frete

  print(f"[{categoria}] - {produto}")
  print(f"Qtd: {quantidade} | Subtotal: R$ {subtotal:.2f} | Final: R$ {valor_final:.2f}")
  print("-" * 40)
```

### Desafio 4

```python
batch_de_transacoes = [
    "PIX, 1500.50, aprovado, 14:30",
    "TED ; 200.00 ; erro ; 10:15",
    {"tipo": "DOC", "valor": 8500.00, "status": "aprovado", "horario": "11:00"},
    "PIX, -50.00, aprovado, 08:00",
    {"tipo": "PIX", "valor": 1500.00, "status": "aprovado", "horario": "02:30"},
    "  TED,   55000.00  , aprovado  , 15:00",
    {"tipo": "BOLETO", "valor": 150.00, "status": "erro", "horario": "19:00"},
    "PIX ; 25000.00 ; aprovado ; 23:15"
]
```

```python
# 1. FUNÇÃO DE LIMPEZA COM HORÁRIO
def limpar_transacao(registro):
    tipo = ""
    valor = 0.0
    status = ""
    hora = 0 # Guarda apenas a hora inteira (ex: 14:30 -> 14)

    if type(registro) == str:
        padrao = registro.replace(";", ",")
        dados = padrao.split(",")

        tipo = dados[0].strip().upper()
        valor = float(dados[1].strip())
        status = dados[2].strip().lower()

        # Pega a hora do formato "14:30" -> pega o "14" e vira int
        horario_str = dados[3].strip()
        hora = int(horario_str.split(":")[0])

    elif type(registro) == dict:
        tipo = registro["tipo"].upper()
        valor = float(registro["valor"])
        status = registro["status"].lower()
        hora = int(registro["horario"].split(":")[0])

    if valor < 0:
        return None

    return {"tipo": tipo, "valor": valor, "status": status, "hora": hora}


# 2. FUNÇÃO DE RISCO (COM REGRA NOTURNA)
def avaliar_risco(valor, tipo, hora):
    # Verifica se é madrugada (entre 22h e 06h)
    eh_madrugada = hora >= 22 or hora < 6

    # Regras Noturnas
    if eh_madrugada:
        if valor > 20000:
            return "CRÍTICO"
        elif tipo == "PIX" and valor > 1000:
            return "ALTO"

    # Regras Diurnas / Gerais
    if valor > 50000:
        return "CRÍTICO"
    elif tipo == "PIX" and valor > 5000:
        return "ALTO"
    elif tipo == "TED" and valor > 10000:
        return "ALTO"
    else:
        return "BAIXO"


# 3. PIPELINE DE DADOS
limpas = [limpar_transacao(t) for t in batch_de_transacoes]
validas = [t for t in limpas if t is not None and t["status"] != "erro"]


# 4. RELATÓRIO FINAL
print("=== ALERTA DE TRANSAÇÕES SUSPEITAS ===")
for t in validas:
    risco = avaliar_risco(t["valor"], t["tipo"], t["hora"])

    if risco in ["ALTO", "CRÍTICO"]:
        print(f"[{risco}] {t['tipo']} de R$ {t['valor']:.2f} às {t['hora']}h (Status: {t['status']})")
```
